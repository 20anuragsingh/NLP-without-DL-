```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
unique_text=df['text'].unique()
emotion_number={}
it=0;
for i in unique_emotion:
  emotion_number[i]=it;
  it+=1;

df["emotion"]=df["emotion"].map(emotion_number)
```

```python
df
df['text']=df['text'].apply( lambda x: x.lower())
```

```python
df
```

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt):
  new=''
  for i in txt:
    if not i.isdigit():
      new +=i
  return new
df['text']=df['text'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
  new=''
  for i in txt:
    if i.isascii():
      new += i
  return new
df['text']= df['text'].apply(remove_emojis)
```

```python
df
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import  word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words= set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words= txt.split()
  cleaned= []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return " ".join(cleaned)
df['text']=df['text'].apply(remove)
```

```python
df
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

bow_vetorizer=CountVectorizer()
x_train_bow=bow_vetorizer.fit_transform(x_train)
x_test_bow=bow_vetorizer.transform(x_test)

nb_model=MultinomialNB()
nb_model.fit(x_train_bow,y_train)

pred_bow=nb_model.predict(x_test_bow)
# accuracy_score(y_test,pred_bow)
print(accuracy_score(y_test,pred_bow))

```

```python
pred_bow
```

```python
from sklearn.linear_model import LogisticRegression

lr_model=LogisticRegression(max_iter=1000)
lr_model.fit(x_train_bow,y_train)

pred_lr=lr_model.predict(x_test_bow)
print(accuracy_score(y_test,pred_lr))
```
